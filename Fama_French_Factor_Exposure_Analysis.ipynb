```python
# ============================================================
# FAMA-FRENCH 5 + MOMENTUM FACTOR EXPOSURE ANALYSIS
# FINAL CLEAN GOOGLE COLAB VERSION
# ============================================================

# ============================================================
# 1. INSTALL
# ============================================================

!pip -q install yfinance statsmodels requests pandas numpy matplotlib


# ============================================================
# 2. IMPORTS
# ============================================================

import io
import re
import zipfile
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import requests
import matplotlib.pyplot as plt
import yfinance as yf
import statsmodels.api as sm

from statsmodels.regression.rolling import RollingOLS
from statsmodels.stats.outliers_influence import variance_inflation_factor

warnings.filterwarnings("ignore")


# ============================================================
# 3. CONFIGURATION
# ============================================================

TICKER = "AAPL"

START_DATE = "2021-01-01"
END_DATE = "2026-08-17"

ROLLING_WINDOW = 252
MIN_ROLLING_OBS = 200
HAC_LAGS = 5

FACTORS = [
    "Mkt-RF",
    "SMB",
    "HML",
    "RMW",
    "CMA",
    "Mom"
]

OUTPUT_DIR = Path("factor_analysis_outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

FF5_URL = (
    "https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/ftp/"
    "F-F_Research_Data_5_Factors_2x3_daily_CSV.zip"
)

MOM_URL = (
    "https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/ftp/"
    "F-F_Momentum_Factor_daily_CSV.zip"
)

print("=" * 80)
print("FAMA-FRENCH 5 + MOMENTUM FACTOR EXPOSURE ANALYSIS")
print("=" * 80)
print()
print("Ticker:", TICKER)
print("Requested period:", START_DATE, "->", END_DATE)
print()


# ============================================================
# 4. DOWNLOAD KENNETH FRENCH FILE
# ============================================================

def download_french_file(url):

    print("Downloading:", url)

    response = requests.get(
        url,
        timeout=(10, 30),
        headers={"User-Agent": "Mozilla/5.0"}
    )

    response.raise_for_status()

    if len(response.content) < 1000:
        raise RuntimeError(
            "The French Data Library returned an unexpectedly small response."
        )

    with zipfile.ZipFile(
        io.BytesIO(response.content)
    ) as archive:

        files = [
            name
            for name in archive.namelist()
            if (
                not name.endswith("/")
                and name.lower().endswith((".csv", ".txt"))
            )
        ]

        if not files:
            raise RuntimeError(
                "No CSV/TXT file was found inside the French ZIP archive."
            )

        raw = archive.read(files[0])

    return raw.decode("latin-1")


# ============================================================
# 5. ROBUST DAILY FRENCH PARSER
# ============================================================

def parse_french_daily(text, expected_columns):

    lines = text.splitlines()

    header_index = None

    # Find a line containing the expected factor names.
    for i, line in enumerate(lines):

        clean = line.strip()

        if not clean:
            continue

        tokens = [
            x.strip()
            for x in re.split(r"[,\s]+", clean)
            if x.strip()
        ]

        token_set = {
            token.lower()
            for token in tokens
        }

        if all(
            column.lower() in token_set
            for column in expected_columns
        ):

            header_index = i
            break

    if header_index is None:

        raise RuntimeError(
            "Could not find the factor header.\n"
            f"Expected columns: {expected_columns}"
        )

    print(
        "Detected header line:",
        lines[header_index]
    )

    # --------------------------------------------------------
    # Parse daily observations.
    #
    # We do NOT rely on whether the source uses commas
    # or spaces.
    # --------------------------------------------------------

    rows = []

    n_expected = len(expected_columns) + 1
    # +1 = Date

    for line in lines[header_index + 1:]:

        clean = line.strip()

        if not clean:
            continue

        parts = [
            x
            for x in re.split(r"[,\s]+", clean)
            if x
        ]

        if not parts:
            continue

        first = parts[0]

        # Daily French observations are YYYYMMDD.
        if (
            len(first) == 8
            and first.isdigit()
        ):

            # We only need the date + expected factor values.
            if len(parts) >= n_expected:

                rows.append(
                    parts[:n_expected]
                )

        elif rows:
            # Stop when the daily table ends.
            break

    if not rows:

        raise RuntimeError(
            "No daily YYYYMMDD observations were found."
        )

    # Explicitly create Date + expected columns.
    columns = ["Date"] + list(expected_columns)

    df = pd.DataFrame(
        rows,
        columns=columns
    )

    # Parse date.
    df["Date"] = pd.to_datetime(
        df["Date"],
        format="%Y%m%d",
        errors="coerce"
    )

    df = df[
        df["Date"].notna()
    ].copy()

    df.set_index(
        "Date",
        inplace=True
    )

    # Convert factors to numeric.
    for column in expected_columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

    # Remove duplicates.
    df = df[
        ~df.index.duplicated(keep="last")
    ]

    df.sort_index(inplace=True)

    # Final structural validation.
    missing = [
        column
        for column in expected_columns
        if column not in df.columns
    ]

    if missing:

        raise RuntimeError(
            "French parser failed.\n"
            f"Missing columns: {missing}\n"
            f"Actual columns: {list(df.columns)}"
        )

    return df


# ============================================================
# 6. FAMA-FRENCH 5 FACTORS
# ============================================================

print()
print("STEP 1 — Loading Fama-French 5 factors")

ff5_text = download_french_file(
    FF5_URL
)

ff5 = parse_french_daily(
    ff5_text,
    [
        "Mkt-RF",
        "SMB",
        "HML",
        "RMW",
        "CMA",
        "RF"
    ]
)

print()
print("FF5 rows:", len(ff5))
print("FF5 columns:", list(ff5.columns))


# ============================================================
# 7. MOMENTUM
# ============================================================

print()
print("STEP 2 — Loading Momentum")

mom_text = download_french_file(
    MOM_URL
)

mom = parse_french_daily(
    mom_text,
    ["Mom"]
)

print()
print("Momentum rows:", len(mom))
print("Momentum columns:", list(mom.columns))


# ============================================================
# 8. COMBINE FACTORS
# ============================================================

print()
print("STEP 3 — Combining factor datasets")

factors = ff5[
    [
        "Mkt-RF",
        "SMB",
        "HML",
        "RMW",
        "CMA",
        "RF"
    ]
].join(
    mom[
        ["Mom"]
    ],
    how="inner"
)

factors = factors.loc[
    START_DATE:END_DATE
].copy()

# French publishes the factor returns in percentage units.
# Convert percentage -> decimal.
factors = factors / 100.0

if factors.empty:

    raise RuntimeError(
        "The combined factor dataframe is empty."
    )


# ============================================================
# FACTOR VALIDATION
# ============================================================

expected_factor_columns = [
    "Mkt-RF",
    "SMB",
    "HML",
    "RMW",
    "CMA",
    "RF",
    "Mom"
]

if list(factors.columns) != expected_factor_columns:

    raise RuntimeError(
        "FACTOR COLUMN VALIDATION FAILED.\n"
        f"Expected: {expected_factor_columns}\n"
        f"Actual: {list(factors.columns)}"
    )

print()
print("=" * 80)
print("FACTOR DATA VALIDATION")
print("=" * 80)

print("Rows:", len(factors))
print("Start:", factors.index.min().date())
print("End:", factors.index.max().date())
print("Columns:", list(factors.columns))

print()
print("Missing values:")

print(
    factors.isna().sum()
)

print()
print("First 5 factor observations:")

display(
    factors.head()
)


# ============================================================
# 9. YAHOO FINANCE
# ============================================================

print()
print("STEP 4 — Downloading Yahoo Finance data")

prices = yf.download(
    TICKER,
    start=START_DATE,
    end=END_DATE,
    auto_adjust=True,
    progress=False,
    threads=False
)

if prices is None or prices.empty:

    raise RuntimeError(
        f"Yahoo Finance returned no data for {TICKER}."
    )


# Handle MultiIndex.
if isinstance(
    prices.columns,
    pd.MultiIndex
):

    if "Close" not in (
        prices.columns
        .get_level_values(0)
    ):

        raise RuntimeError(
            "Yahoo response does not contain Close prices."
        )

    close = prices["Close"]

    if isinstance(
        close,
        pd.DataFrame
    ):

        if TICKER in close.columns:

            close = close[TICKER]

        else:

            close = close.iloc[:, 0]

else:

    if "Close" not in prices.columns:

        raise RuntimeError(
            "Yahoo response does not contain Close prices."
        )

    close = prices["Close"]


if isinstance(
    close,
    pd.DataFrame
):

    close = close.iloc[:, 0]


close = pd.Series(
    close,
    name=TICKER
)

close.index = pd.to_datetime(
    close.index
)

if close.index.tz is not None:

    close.index = (
        close.index
        .tz_localize(None)
    )

close.sort_index(inplace=True)

print()
print("Yahoo rows:", len(close))
print("Yahoo start:", close.index.min().date())
print("Yahoo end:", close.index.max().date())


# ============================================================
# 10. STOCK RETURNS
# ============================================================

print()
print("STEP 5 — Calculating stock returns")

stock_returns = close.pct_change(
    fill_method=None
)

# Mask clearly erroneous observations.
stock_returns = stock_returns.mask(
    stock_returns.abs() > 0.95
)


# ============================================================
# 11. MERGE STOCK + FACTORS
# ============================================================

print()
print("STEP 6 — Merging stock returns with factors")

data = pd.DataFrame({
    "Stock_Return": stock_returns
}).join(
    factors,
    how="inner"
)

print(
    "Rows after date intersection:",
    len(data)
)

print()
print("Missing values before cleaning:")

display(
    data.isna().sum().to_frame("Missing")
)

data = data.dropna(
    how="any"
)

if data.empty:

    raise RuntimeError(
        "No overlapping observations remain after cleaning.\n\n"
        f"Yahoo: {close.index.min().date()} -> "
        f"{close.index.max().date()}\n"
        f"French: {factors.index.min().date()} -> "
        f"{factors.index.max().date()}"
    )


# ============================================================
# 12. EXCESS RETURNS
# ============================================================

data["Excess_Return"] = (
    data["Stock_Return"]
    - data["RF"]
)

data = data[
    [
        "Stock_Return",
        "Excess_Return",
        "Mkt-RF",
        "SMB",
        "HML",
        "RMW",
        "CMA",
        "Mom",
        "RF"
    ]
]

print()
print("=" * 80)
print("MODEL DATASET")
print("=" * 80)

print(
    "Observations:",
    len(data)
)

print(
    "Start:",
    data.index.min().date()
)

print(
    "End:",
    data.index.max().date()
)

display(
    data.head()
)


# ============================================================
# 13. STATIC SIX-FACTOR REGRESSION
# ============================================================

print()
print("STEP 7 — Running static six-factor OLS")

X = sm.add_constant(
    data[FACTORS],
    has_constant="add"
)

y = data["Excess_Return"]

results = sm.OLS(
    y,
    X
).fit(
    cov_type="HAC",
    cov_kwds={
        "maxlags": HAC_LAGS
    },
    use_t=True
)

regression_table = pd.DataFrame({
    "Coefficient": results.params,
    "HAC_Std_Error": results.bse,
    "t_stat": results.tvalues,
    "p_value": results.pvalues,
    "Significant_5pct": results.pvalues < 0.05,
    "Significant_1pct": results.pvalues < 0.01
})

print()
print("=" * 80)
print("FACTOR EXPOSURES")
print("=" * 80)

display(
    regression_table.round(4)
)

print()
print("FULL OLS SUMMARY")

print(
    results.summary()
)


# ============================================================
# 14. ROLLING REGRESSION
# ============================================================

print()
print("STEP 8 — Running rolling 252-day regression")

if len(data) < ROLLING_WINDOW:

    raise RuntimeError(
        f"Need at least {ROLLING_WINDOW} observations "
        f"for rolling regression. "
        f"Available: {len(data)}"
    )

rolling_X = sm.add_constant(
    data[FACTORS],
    has_constant="add"
)

rolling_model = RollingOLS(
    endog=data["Excess_Return"],
    exog=rolling_X,
    window=ROLLING_WINDOW,
    min_nobs=MIN_ROLLING_OBS,
    missing="drop"
)

rolling_results = rolling_model.fit(
    method="pinv",
    params_only=True
)

rolling = rolling_results.params.copy()

print(
    "✅ Rolling regression complete."
)


# ============================================================
# 15. VIF
# ============================================================

print()
print("STEP 9 — Calculating factor VIF")

vif_X = sm.add_constant(
    data[FACTORS],
    has_constant="add"
)

vif_rows = []

for i, factor in enumerate(
    vif_X.columns
):

    if factor == "const":
        continue

    try:

        vif_value = variance_inflation_factor(
            vif_X.values,
            i
        )

    except Exception:

        vif_value = np.inf

    vif_rows.append({
        "Factor": factor,
        "VIF": vif_value
    })

vif_table = pd.DataFrame(
    vif_rows
)

display(
    vif_table.round(3)
)


# ============================================================
# 16. PERFORMANCE METRICS
# ============================================================

print()
print("STEP 10 — Calculating performance metrics")

returns = data["Stock_Return"]
excess = data["Excess_Return"]

wealth = (
    1 + returns
).cumprod()

years = len(returns) / 252.0

if (
    years > 0
    and wealth.iloc[-1] > 0
):

    CAGR = (
        wealth.iloc[-1]
        ** (1 / years)
    ) - 1

else:

    CAGR = np.nan

annualized_volatility = (
    returns.std(ddof=1)
    * np.sqrt(252)
)

if excess.std(ddof=1) > 0:

    sharpe = (
        excess.mean()
        / excess.std(ddof=1)
    ) * np.sqrt(252)

else:

    sharpe = np.nan

drawdown = (
    wealth / wealth.cummax()
) - 1

maximum_drawdown = drawdown.min()

annualized_alpha = (
    results.params["const"]
    * 252
)

metrics = pd.Series({
    "Observations": len(data),
    "CAGR": CAGR,
    "Annualized Volatility": annualized_volatility,
    "Sharpe Ratio": sharpe,
    "Maximum Drawdown": maximum_drawdown,
    "R-Squared": results.rsquared,
    "Adjusted R-Squared": results.rsquared_adj,
    "Annualized Alpha": annualized_alpha,
    "Alpha t-stat": results.tvalues["const"],
    "Alpha p-value": results.pvalues["const"]
})

display(
    metrics.to_frame("Value").round(6)
)


# ============================================================
# 17. FACTOR ATTRIBUTION
# ============================================================

print()
print("STEP 11 — Factor attribution")

annual_factor_returns = (
    data[FACTORS].mean()
    * 252
)

factor_contribution = (
    results.params[FACTORS]
    * annual_factor_returns
)

attribution = pd.Series({
    "Alpha":
        results.params["const"] * 252,

    "Market":
        factor_contribution["Mkt-RF"],

    "Size":
        factor_contribution["SMB"],

    "Value":
        factor_contribution["HML"],

    "Profitability":
        factor_contribution["RMW"],

    "Investment":
        factor_contribution["CMA"],

    "Momentum":
        factor_contribution["Mom"]
})

actual_excess_return = (
    data["Excess_Return"].mean()
    * 252
)

attribution["Residual"] = (
    actual_excess_return
    - attribution.sum()
)

display(
    attribution.to_frame(
        "Annualized Contribution"
    ).round(6)
)


# ============================================================
# 18. CHART — ROLLING FACTOR EXPOSURES
# ============================================================

print()
print("STEP 12 — Rolling exposure charts")

for factor in FACTORS:

    beta = (
        rolling[factor]
        .dropna()
    )

    if beta.empty:
        continue

    fig, ax = plt.subplots(
        figsize=(11, 5)
    )

    ax.plot(
        beta.index,
        beta.values,
        linewidth=1.5
    )

    ax.axhline(
        0,
        linestyle="--",
        linewidth=1
    )

    ax.set_title(
        f"Rolling {ROLLING_WINDOW}-Day {factor} Exposure",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Date")
    ax.set_ylabel("Beta")

    ax.grid(alpha=0.25)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()
    plt.show()
    plt.close(fig)


# ============================================================
# 19. CHART — ALPHA SIGNIFICANCE
# ============================================================

alpha_annualized = (
    results.params["const"]
    * 252
)

alpha_se_annualized = (
    results.bse["const"]
    * 252
)

fig, ax = plt.subplots(
    figsize=(7, 5)
)

ax.errorbar(
    ["Alpha"],
    [alpha_annualized],
    yerr=np.array([
        [1.96 * alpha_se_annualized],
        [1.96 * alpha_se_annualized]
    ]),
    fmt="o",
    capsize=7,
    markersize=8
)

ax.axhline(
    0,
    linestyle="--",
    linewidth=1
)

ax.set_title(
    "Annualized Alpha - 95% HAC Confidence Interval",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel(
    "Annualized Alpha"
)

ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
plt.close(fig)


# ============================================================
# 20. CHART — FACTOR ATTRIBUTION
# ============================================================

fig, ax = plt.subplots(
    figsize=(11, 5)
)

ax.bar(
    attribution.index,
    attribution.values
)

ax.axhline(
    0,
    linewidth=1
)

ax.set_title(
    "Annualized Factor Attribution",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel(
    "Return Contribution"
)

plt.xticks(
    rotation=25
)

ax.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()
plt.close(fig)


# ============================================================
# 21. CHART — REALIZED VS FACTOR MODEL
# ============================================================

prediction_X = sm.add_constant(
    data[FACTORS],
    has_constant="add"
)

fitted = pd.Series(
    results.predict(prediction_X),
    index=data.index
)

realized_curve = (
    1 + data["Excess_Return"]
).cumprod()

fitted_curve = (
    1 + fitted
).cumprod()

fig, ax = plt.subplots(
    figsize=(11, 5)
)

ax.plot(
    realized_curve.index,
    realized_curve.values,
    label="Realized",
    linewidth=1.6
)

ax.plot(
    fitted_curve.index,
    fitted_curve.values,
    label="Factor Model",
    linewidth=1.6,
    linestyle="--"
)

ax.set_yscale("log")

ax.set_title(
    "Realized vs Factor-Model Excess Return",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel(
    "Growth of $1 - Log Scale"
)

ax.legend(frameon=False)

ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
plt.close(fig)


# ============================================================
# 22. SAVE ALL OUTPUTS
# ============================================================

print()
print("STEP 13 — Saving research outputs")

data.to_csv(
    OUTPUT_DIR / "model_dataset.csv"
)

regression_table.to_csv(
    OUTPUT_DIR / "regression_statistics.csv"
)

rolling.to_csv(
    OUTPUT_DIR / "rolling_exposures.csv"
)

vif_table.to_csv(
    OUTPUT_DIR / "vif_diagnostics.csv",
    index=False
)

metrics.to_csv(
    OUTPUT_DIR / "performance_metrics.csv",
    header=["Value"]
)

attribution.to_csv(
    OUTPUT_DIR / "factor_attribution.csv",
    header=["Annualized Contribution"]
)

with open(
    OUTPUT_DIR / "ols_summary.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        results.summary().as_text()
    )


# ============================================================
# 23. FINAL SUMMARY
# ============================================================

print()
print("=" * 80)
print("✅ PROJECT COMPLETE")
print("=" * 80)

print()
print("Ticker:", TICKER)
print("Observations:", f"{len(data):,}")

print(
    "Sample:",
    data.index.min().date(),
    "->",
    data.index.max().date()
)

print(
    "CAGR:",
    f"{CAGR:.2%}"
)

print(
    "Annualized Volatility:",
    f"{annualized_volatility:.2%}"
)

print(
    "Sharpe Ratio:",
    f"{sharpe:.3f}"
)

print(
    "Maximum Drawdown:",
    f"{maximum_drawdown:.2%}"
)

print(
    "R-Squared:",
    f"{results.rsquared:.4f}"
)

print(
    "Annualized Alpha:",
    f"{annualized_alpha:.2%}"
)

print(
    "Alpha t-stat:",
    f"{results.tvalues['const']:.3f}"
)

print(
    "Alpha p-value:",
    f"{results.pvalues['const']:.4f}"
)

print()
print("Output directory:")
print(
    OUTPUT_DIR.resolve()
)

print()
print("Generated files:")

for file in sorted(
    OUTPUT_DIR.iterdir()
):

    print(
        " -",
        file.name
    )
```
